```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("/Users/ravihw18/ML-Algorithms/10-Clustering/10.6-Isolation_Forest/thyroid_dataset.csv")
```

```python
X=df.drop("Outlier_label", axis=1)
y=df["Outlier_label"]
```

```python
scaler= StandardScaler()
X_scaled= scaler.fit_transform(X)
```

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
lof=LocalOutlierFactor(contamination=0.036)

labels=lof.fit_predict(X_scaled)
```

```python

from sklearn.decomposition import PCA
pca =PCA(
    n_components=2
)

X_pca=pca.fit_transform(X_scaled)
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0],X_pca[:,1], c=labels)
plt.xlabel("PCA1")
plt.ylabel("PCA2")
```

```python
# Printing no. of outliers and normal points

import numpy as np
n_outlier= np.sum(labels==-1)
n_normal=np.sum(labels==1)

print("Outliers:",n_outlier )
print("n_normal:",n_normal )
```
